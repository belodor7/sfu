```python
// Наследование
class Person{
    private string _name = "";
    public string Name
    {
        get{ return _name; }
        set{ _name = value; }
    }
    public void Print()
    {
        Console.WriteLine(Name);
    }
}

class Employee : Person
{
    public void PrintName()
    {
        Console.WriteLine(_name); // так не получится! _name - private
    }
}
```

Employee наследует все свойства, методы и поля из Person. Не передаются только конструкторы.

Ограничения:
1. Не поддержвиеося множественное наследование. Наследовать можно только один класс
2. Доступ к классу должен быть одинаковый. private - private; public - public; НО НЕ! public - private (или наоборот).
3. Если классы находятся в разных сборках, то можно наследовать только public классы.
